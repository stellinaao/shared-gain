```python
from sg.models import Encoder, StrategyEncoder

"""test if unbalanced number of trials is correct"""

subj_id = "MR82"
sess_id = "20251027_152036"

encoder = Encoder(subj_id, sess_id)
encoder.fit_encoder()

encoder_mb = StrategyEncoder(
    subj_id, sess_id, norm=False, strategy_filter="mb", balance_strategy=False
)
encoder_mf = StrategyEncoder(
    subj_id, sess_id, norm=False, strategy_filter="mf", balance_strategy=False
)

encoder_mb.fit_encoder()
encoder_mf.fit_encoder()

assert (encoder_mb.num_trials == (encoder.trial_data["strategy"] == 1).sum()) & (
    encoder_mf.num_trials == (encoder.trial_data["strategy"] == -1).sum()
)
```
