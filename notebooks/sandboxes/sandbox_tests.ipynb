```python
from sg.models import Encoder, StrategyEncoder

"""test if unbalanced number of trials is correct"""

subj_id = "MR82"
sess_id = "20251027_152036"

encoder = Encoder(subj_id, sess_id)
encoder.fit_encoder()

encoder_mb = StrategyEncoder(
    subj_id, sess_id, norm=False, strategy_filter="mb", balance_strategy=False
)
encoder_mf = StrategyEncoder(
    subj_id, sess_id, norm=False, strategy_filter="mf", balance_strategy=False
)

encoder_mb.fit_encoder()
encoder_mf.fit_encoder()

assert (encoder_mb.num_trials == (encoder.trial_data["strategy"] == 1).sum()) & (
    encoder_mf.num_trials == (encoder.trial_data["strategy"] == -1).sum()
)
```

## tre

```python
from sg.models import make_tre, Encoder, StrategyEncoder

encoder = make_tre(Encoder, tr_type="dme")(
    subj_id,
    sess_id,
    norm=True,
    tv_keys=["response", "rewarded", "response_prev", "rewarded_prev"],
    stepsize_s=0.1,
)
encoder.verify()

encoder_mb = make_tre(StrategyEncoder, tr_type="dme")(
    subj_id,
    sess_id,
    norm=False,
    stepsize_s=0.1,
    strategy_filter="mb",
)
encoder_mb.verify()

encoder_mf = make_tre(StrategyEncoder, tr_type="dme")(
    subj_id,
    sess_id,
    norm=False,
    stepsize_s=0.1,
    strategy_filter="mf",
)
encoder_mf.verify()
```

```python
assert (
    encoder.trial_data["strategy"].iloc[encoder_mb.idxs].unique() == 1
    and encoder.trial_data["strategy"].iloc[encoder_mf.idxs].unique() == -1
)
```
