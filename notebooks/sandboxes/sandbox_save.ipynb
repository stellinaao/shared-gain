## save

```python
from core.data import subject_ids, session_ids
from sg.models import Encoder
from utils.paths import MODELS_DIR
import pickle
import numpy as np

for subj_id in ["MR82", "MR83"]:
    for sess_id in session_ids[np.where(subject_ids == subj_id)[0][0]]:
        encoder = Encoder(subj_id, sess_id)

        encoder.fit_baseline()
        encoder.fit_encoder()

        encoder.baseline_predict()
        encoder.encoder_predict()

        encoder.get_r2()

        save_path = MODELS_DIR / "encoders" / subj_id / sess_id
        save_path.mkdir(parents=True, exist_ok=True)

        with open(save_path / "encoder.pkl", "wb") as f:
            pickle.dump(encoder, f)
```

## load

```python
encoders = {}
for subj_id in ["MR82", "MR83"]:
    encoders[subj_id] = {}
    for sess_id in session_ids[np.where(subject_ids == subj_id)[0][0]]:
        fpath = MODELS_DIR / "encoders" / subj_id / sess_id

        with open(fpath / "encoder.pkl", "rb") as f:
            encoders[subj_id][sess_id] = pickle.load(f)
```

```python
for subj_id in ["MR82", "MR83"]:
    for sess_id in session_ids[np.where(subject_ids == subj_id)[0][0]]:
        assert (
            encoders[subj_id][sess_id].subj_id == subj_id
            and encoders[subj_id][sess_id].sess_id == sess_id
        )
```
